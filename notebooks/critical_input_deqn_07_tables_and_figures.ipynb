# Critical-input DEQN: tables and figures

Run this after the relevant experiments have been trained and postprocessed. Missing experiment folders are skipped, so the notebook can be run after baseline only and re-run after counterfactuals/sensitivity variants are added.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if not (ROOT / 'src').exists():
    ROOT = ROOT.parent

BASE_ROOT = ROOT / 'baseline_artifacts' / 'critical_input_deqn'
TABLES = BASE_ROOT / 'tables'
FIGURES = BASE_ROOT / 'figures'
POLICY = 'ba'
print(ROOT)
print(TABLES)
print(FIGURES)
```

```python
cmd = [
    sys.executable, '-m', 'src.critical_input_deqn.make_tables',
    '--base-root', str(BASE_ROOT),
    '--output-dir', str(TABLES),
    '--policy', POLICY,
]
subprocess.run(cmd, cwd=ROOT, check=True)
```

```python
cmd = [
    sys.executable, '-m', 'src.critical_input_deqn.make_figures',
    '--base-root', str(BASE_ROOT),
    '--output-dir', str(FIGURES),
    '--policy', POLICY,
]
subprocess.run(cmd, cwd=ROOT, check=True)
```

```python
print('Tables:')
for p in sorted(TABLES.glob('*')):
    print(' ', p.name)
print('Figures:')
for p in sorted(FIGURES.glob('*')):
    print(' ', p.name)
```
