# Critical-input DEQN: tables and figures

Run this after the relevant experiments have been trained and postprocessed. Missing experiment folders are skipped, so the notebook can be run after baseline only and re-run after counterfactuals/sensitivity variants are added.

```python
# Configure output folders for final tables and figures.
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if not (ROOT / 'src').exists():
    ROOT = ROOT.parent

BASE_ROOT = ROOT / 'baseline_artifacts' / 'critical_input_deqn'
TABLES = BASE_ROOT / 'tables'
FIGURES = BASE_ROOT / 'figures'
POLICY = 'ba'
print(ROOT)
print(TABLES)
print(FIGURES)

# Stream subprocess output line by line in Colab instead of waiting silently.
def run_stream(cmd, *, cwd=ROOT, env=None):
    print('Running:', ' '.join(map(str, cmd)), flush=True)
    proc = subprocess.Popen(
        cmd,
        cwd=cwd,
        env=env,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
    )
    for line in proc.stdout:
        print(line, end='', flush=True)
    ret = proc.wait()
    if ret != 0:
        raise subprocess.CalledProcessError(ret, cmd)

```

```python
# Build calibration, moment, counterfactual, and sensitivity tables.
cmd = [
    sys.executable, '-u', '-m', 'src.critical_input_deqn.make_tables',
    '--base-root', str(BASE_ROOT),
    '--output-dir', str(TABLES),
    '--policy', POLICY,
]
run_stream(cmd, cwd=ROOT)
```

```python
# Build mechanism, policy-comparison, sensitivity, and distribution figures.
cmd = [
    sys.executable, '-u', '-m', 'src.critical_input_deqn.make_figures',
    '--base-root', str(BASE_ROOT),
    '--output-dir', str(FIGURES),
    '--policy', POLICY,
]
run_stream(cmd, cwd=ROOT)
```

```python
# Print generated table and figure files.
print('Tables:')
for p in sorted(TABLES.glob('*')):
    print(' ', p.name)
print('Figures:')
for p in sorted(FIGURES.glob('*')):
    print(' ', p.name)
```
